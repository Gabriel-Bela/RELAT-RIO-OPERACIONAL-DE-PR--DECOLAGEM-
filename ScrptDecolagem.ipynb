```python
#DEFINIÇÃO INICIAL DAS VARIAVÉIS DOS SENSORES DA NAVE:

temp_interna : float   # Temperatura Interna         (0 °C a 40°C)
temp_externa : float   # Temperatura Externa         (-20 °C a 50°C)
integridade : int      # Integridade Estrutural      (1 = Íntegra)
energia : int          # Níveis de Energia           (>=  70% e <= 100%)
press : float          # Pressão dos Tanques         (80% a 100%)
modulos : int          # Status dos Módulos Críticos (Todos em "OK")

#VARIAVÉIS DA ANÁLISE ENERGÉTICA:
cap = float (200)      # Capacidade de Energia da Nave (200 kWh)
carga: float           # Carga de Energia da Nave (Variavél duplicada)
perdas = float(10)     # Percentual de perdas Energéticas (10%)
consumo_decolagem = float(60)   # Consumo da Decolagem
```

```python
#Recebimento dos dados da nave
temp_interna = float(input("Qual é a Temperatura Interna em °C?"))
temp_externa = float(input("Qual é a Temperatura Externa em °C?"))
integridade  = int(input("Qual a Integridade Estrutural?\nDigite (1)-Integro  ou  (0)-Não Integro"))
energia      = int(input("Qual é o nível de Energia?"))
press        = float(input("Qual é a pressão dos tanques?"))
modulos      = int(input("Qual o Status dos módulos críticos?\nDigite (1)-Bom  ou  (0)-Ruim"))
```

```python
#verificação das condições de decolagem:
print ("                        === SISTEMA DE VERIFICAÇÃO DA AURORA-1 ===\n")

#Verificação da temperatura interna
if temp_interna >= 0 and temp_interna <=40:
    print("Temperatura Interna OK!✅")
    temp_interna= bool(True)                #Colocado status verdadeiro para decolagem.
elif temp_interna < 0:                              
    print("Temperatura Interna Muito Baixa - Compromete a Decolagem!❌") 
else: print("Temperatura Interna Muito Alta - Compromete a Decolagem!❌")

#Verificação da temperatura externa
if temp_externa >= -20 and temp_externa <=50:
    print("Temperatura Externa OK!✅")
    temp_externa = bool(True)               #Colocado status verdadeiro para decolagem.
elif temp_externa < -20:                             
    print("Temperatura Externa Muito Baixa - Compromete a Decolagem!❌")   
else: print("Temperatura Externa Muito Alta - Compromete a Decolagem!❌")

#Verificação da integridade
if integridade != 0 and integridade != 1:
    integridade = int(input("Valor inválido para integridade estrutural. Por favor, insira 0 ou 1."))
if integridade == 1:
    print("Integridade OK!✅")
    integridade = bool(True)              #Colocado status verdadeiro para decolagem.
else: print("Integridade não satisfatótia - Compromete a Decolagem!❌")

#Verificação dos níveis de energia
if energia >= 70 and energia <=100:
    print ("Niveis de energia OK!✅")
    carga = float(energia)
    energia = bool(True)                  #Colocado status verdadeiro para decolagem.                                   
elif energia < 70:
    print(f"O nível de {energia}% de energia não é o suficiente para decolar.❌")
else: print(f"O nível da energia({energia}%) está muito alto. Por favor, verificar os sensores.❌")

#Verificação da Pressão dos tanques
if press >= 80 and press <= 100:
    print("Pressão dos Tanques OK!✅")
    press= bool(True)                       #Colocado status verdadeiro para decolagem. 
elif press<80:
    print(f"A pressão dos tanques ({press}%) está muito baixa - Compromete a Decolagem!❌")
else: print(f"A pressão dos tanques ({press}%) está muito alta - Compromete a Decolagem!❌")

#Verificação dos módulos críticos
if modulos != 0 and modulos != 1:
    modulos = int(input("Valor inválido para status dos módulos críticos. Por favor, insira 0 ou 1."))
if modulos == 1:
    print("Status dos Módulos Críticos OK!✅")       
    modulos= bool (True)                  #Colocado status verdadeiro para decolagem.
else: print("Os Módulos Críticos não estão operacionais - Compromet a Decolagem!❌")

#Analisar dados, após analise verificar se pode Decolar ou abortar a Decolagem

if temp_interna==True and temp_externa==True and integridade==True and energia==True and press==True and modulos==True:
    print("\n            🚀DECOLAGEM AUTORIZADA!🚀 Todos os sistemas estão operacionais!\n")

    #Análise Energética após a decolagem (Colocar na Decolagem)
    input("Pressione Enter para iniciar a análise energética após a decolagem...")

    print("\n\nPainel de análise energética após a decolagem:\n")

    # Energia inicial da nave
    energia_inicial = cap * (carga / 100)               

    # Cálculo das perdas energéticas
    energia_perdas = energia_inicial * (perdas / 100)   

    # Energia restante após a decolagem
    energia_restante = energia_inicial - energia_perdas - consumo_decolagem
    
    #Rendimento energético da decolagem
    rendimento = (consumo_decolagem / (consumo_decolagem + energia_perdas)) * 100

    # Percentual de energia restante
    percentual_restante = (energia_restante / energia_inicial) * 100

    print(f"Energia inicial da nave:                {energia_inicial:.2f} kWh") #
    print(f"Consumo de energia na decolagem:        {consumo_decolagem:.2f} kWh") #
    print(f"Perdas energéticas durante a decolagem: {energia_perdas:.2f} kWh")
    print(f"Energia restante:                       {energia_restante:.2f} kWh")
    print(f"Rendimento energético na decolagem:     {rendimento:.2f}%")
    print(f"\nStatus atual da energia:                {percentual_restante:.2f}%")

#Caso algum dos sensores não esteja dentro dos parâmetros, a decolagem será abortada.
else: print("\n                 🚫DECOLAGEM ABORTADA!🚫")

print("Fim.")


```
